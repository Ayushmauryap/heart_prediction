```python
import pandas as pd
import numpy as np
import seaborn as sns


from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('heart.csv')
```

```python
df.sample(10)
```

```python
df.isnull().sum()
```

```python
df.sample(3)
```

```python
train_X,test_X,train_y,test_y = train_test_split(df.drop(columns=['HeartDisease']),df['HeartDisease'],test_size=0.2,random_state=42)
```

```python
train_X.info()
```

```python
trf1 =ColumnTransformer([
    ('oe_ChestPainType',OrdinalEncoder(categories=[['ASY','NAP','ATA','TA']]),[2]),
    ('oe_RestingECG',OrdinalEncoder(categories=[['Normal','ST','LVH']]),[6]),
    ('oe_ExerciseAngina',OrdinalEncoder(categories=[['Y','N']]),[8]),
    ('oe_ST_Slop',OrdinalEncoder(categories=[['Flat','Up','Down']]),[9]),
    ('oe_sex',OrdinalEncoder(categories=[['M','F']]),[1])
],remainder='passthrough')
```

```python
#Scaling
trf3=MinMaxScaler()
```

```python
#feature Selection
trf4=SelectKBest(score_func=chi2,k=10)
```

```python
#train the model
trf5=DecisionTreeClassifier(random_state=42)
```

```python
pipe = Pipeline([
    ('trf1',trf1),
    ('trf3',trf3),
    ('trf4',trf4),
    ('trf5',trf5)
])
```

```python
pipe.fit(train_X,train_y)
```

```python
pipe.named_steps
```

```python
from sklearn import set_config
```

```python
set_config(display='diagram')
```

```python
y_pred = pipe.predict(test_X)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(test_y,y_pred)
```

```python
#cross validation using cross_val_score
from sklearn.model_selection import cross_val_score
cross_val_score(pipe,train_X,train_y,cv=5,scoring='accuracy').mean()
```
